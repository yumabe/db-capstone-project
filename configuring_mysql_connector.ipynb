```python
import mysql.connector as connector
from mysql.connector import Error

# --- PASO 1: Conexión---
db_config = {
    "user": "yumabe",
    "password": "yumabe",  
    "host": "127.0.0.1",
    "db": "littlelemondm"  
}

try:
    # Establecemos la conexión
    connection = connector.connect(**db_config)
    print(f"Conexión a la base de datos '{db_config['db']}' establecida.")

    # --- INICIO DE LA TAREA 2 ---

    # Paso 2: Crear un cursor
    cursor = connection.cursor()
    print("Cursor creado.")

    # Paso 3: Definir la consulta SQL
    show_tables_query = "SHOW TABLES"

    # Paso 4: Ejecutar la consulta
    cursor.execute(show_tables_query)
    print("Consulta 'SHOW TABLES' ejecutada.")

    # Paso 5: Recuperar todos los resultados
    results = cursor.fetchall()
    print("Resultados obtenidos con fetchall().")
    
    # Paso 6: Imprimir los resultados
    print("\n------ Tablas en la base de datos ------")
    print(results)
    print("------------------------------------")

    # Opcional: Imprimir de forma más legible
    print("\nLista de tablas:")
    for table in results:
        print(f"- {table[0]}")
    
    # --- FIN DE LA TAREA 2 ---

except Error as e:
    print(f"Error durante la ejecución: {e}")

finally:
    # Cerramos la conexión SIEMPRE
    if 'connection' in locals() and connection.is_connected():
        cursor.close()
        connection.close()
        print("\nCursor y conexión cerrados correctamente.")
```

```python
import mysql.connector as connector
from mysql.connector import Error

# --- Configuración de la Conexión ---
db_config = {
    "user": "yumabe",
    "password": "yumabe",
    "host": "127.0.0.1",
    "db": "littlelemondm"
}

try:
    # Establecemos la conexión
    connection = connector.connect(**db_config)
    cursor = connection.cursor()

    # --- INICIO DE LA TAREA 3 ---

    promo_campaign_query = """
    SELECT
        c.customername,
        c.contactnumber,
        c.email,
        o.totalcost
    FROM
        customers AS c
    JOIN
        orders AS o ON c.customerid = o.customerid
    WHERE
        o.totalcost > 60;
    """

    # Ejecutamos la consulta
    cursor.execute(promo_campaign_query)
    print("Consulta para la campaña promocional ejecutada.")

    # Recuperamos todos los resultados
    results = cursor.fetchall()

    # Imprimimos los resultados de forma clara
    print("\n------ Clientes con órdenes mayores a $60 ------")
    if not results:
        print("No se encontraron clientes que cumplan con el criterio.")
    else:
        # Imprimimos una cabecera para que la tabla se entienda
        print(f"{'Nombre Completo':<25} | {'Contacto':<15} | {'Email':<25} | {'Costo Total':<10}")
        print("-" * 80)
        # Iteramos sobre cada fila (cliente) en los resultados
        for row in results:
            full_name, contact, email, total_cost = row
            print(f"{full_name:<25} | {contact:<15} | {email:<25} | ${total_cost:<9.2f}")
    
    print("---------------------------------------------")

    # --- FIN DE LA TAREA 3 ---

except Error as e:
    print(f"Error durante la ejecución: {e}")

finally:
    # Cerramos la conexión SIEMPRE
    if 'connection' in locals() and connection.is_connected():
        cursor.close()
        connection.close()
        print("\nCursor y conexión cerrados correctamente.")
```
